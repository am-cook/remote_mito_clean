```r
# we will make 20 synthetic sequences. 
# in total there will be 10 possible integrations within each sequence
# each integration will have length 50
# we will downsample integrations in each sequence so that no (in the first) or some (in the second)
# cells have all 10 integrations
# run the fasta through muscle to see if it can recapitulate the full 10 integration (length 500) seq

```

```r
base_integration_sequence <- sample(c('A', 'C', 'G', 'T'), size = 50, replace = TRUE)
```

```r
set.seed(42)
integration_recovery_prob <- 0.9
max_num_integrations <- 10
num_muts_per_fullseq <- 30
base_integration_sequence <- sample(c('A', 'C', 'G', 'T'), size = 50, replace = TRUE)
cell_names <- paste0('Cell_', seq(1:20))

# for each cell:
# determine number of recovered ints based on max num intes and int recovery prob
# concatenate the base_integration_sequence num_recovered_ints times
# randomly sample num_muts_per_fullseq positions in the concatenated sequence to mutate
full_cell_seqs <- lapply(1:length(cell_names), function(x){
    
    num_recovered_ints <- rbinom(n = 1, size = max_num_integrations, 
                                 prob = integration_recovery_prob)
    this_cell_seq <- rep(base_integration_sequence, num_recovered_ints)

    mutation_positions <- sample(seq(1,length(this_cell_seq)), size = num_muts_per_fullseq, 
                                 replace = FALSE)
    new_bases <- sample(c('A', 'C', 'G', 'T'), size = num_muts_per_fullseq, replace = TRUE)

    for(i in 1:length(mutation_positions)){
        this_cell_seq[i] <- new_bases[i]
    }

    
    return(this_cell_seq)
})

names(full_cell_seqs) <- cell_names
```

```r
write.fasta(full_cell_seqs, names = cell_names, file.out = './different_length_seqs.fasta')
```
